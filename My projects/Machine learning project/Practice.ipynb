```python
#First method direct asking
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
music_df = pd.read_csv("music.csv")
X = music_df.drop(columns=["genre"])
y = music_df["genre"]
model = DecisionTreeClassifier()
model.fit(X, y)
predictions = model.predict([[21,1],[22,0]])
predictions
```

```python
#Calculating the accuracy
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection  import train_test_split
from sklearn.metrics import accuracy_score
music_df = pd.read_csv("music.csv")
X = music_df.drop(columns=["genre"])
y = music_df["genre"]
X_train, X_test,y_train, y_test = train_test_split(X, y, test_size=0.2 ) # tuple with 4 argument
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
predictions = model.predict(X_test)
accuracy_percentage = accuracy_score(y_test, predictions)
accuracy_percentage

```

```python
#Model persisting that is to store our trained model in a file so that we won't be training it over and over when ever we need it
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
import joblib #Changed from "from sklearn.externals import joblib"
music_df = pd.read_csv("music.csv")
X = music_df.drop(columns=["genre"])
y = music_df["genre"]
model = DecisionTreeClassifier()
model.fit(X, y)
joblib.dump(model, "music-recommender-model")
#predictions = model.predict([[21,1]])
```

```python
#Using our stored model to predict
joblib.load('music-recommender-model')
predictions = model.predict([[21,1]])
predictions
```

```python
#Visualisaton of decision tree
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
music_df = pd.read_csv("music.csv")
X = music_df.drop(columns=["genre"])
y = music_df["genre"]
model = DecisionTreeClassifier()
model.fit(X, y)
tree.export_graphviz(model, out_file="music-RecTree.dot",
                     feature_names=["age","gender"], 
                     class_names=sorted(y.unique()),
                     label="all",rounded=True,filled=True)
```
